```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report
import joblib
```

```python
df = pd.read_csv("cleandata.csv",index_col=0)
df.head()
```

```python
#feature engineering

#rating gap which shows squads consistancy
df["rating_range"] = df["squad_max_rating"] - df["squad_min_rating"]
```

```python
#final feature set

feature_cols = [
    "squad_avg_rating",
    "squad_min_rating",
    "squad_rating_std",
    "rating_range",
    "buildUpPlayPassing",
    "defencePressure",
    "chanceCreationShooting",
    "played",
    "wins",
    "losses",
    "draws"
]
```

```python
X = df[feature_cols]
y = df["tier"]
```

```python
#encoding tier to labels

le = LabelEncoder()
y_enc = le.fit_transform(y)
```

```python
#train test split

X_train, X_test, y_train, y_test = train_test_split(
    X, y_enc, test_size=0.2, random_state=42, stratify=y_enc
)
```

```python
#scaling features(very crutial for the neural netwroks part especially)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#defining the models

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "Neural Network (MLP)": MLPClassifier(hidden_layer_sizes=(32, 16),max_iter=2000, random_state=42),
}
```

```python
#training each model
results = {}
 
for name, model in models.items():
    
    if name in ["Logistic Regression", "Neural Network (MLP)"]:
        model.fit(X_train_scaled, y_train)
        preds = model.predict(X_test_scaled)
    else:
        model.fit(X_train, y_train)
        preds = model.predict(X_test)

    acc = accuracy_score(y_test, preds)
    results[name] = acc
 
    print(f"Accuracy: {acc:.4f}")
    print(classification_report(y_test, preds, target_names=le.classes_))
 
```

```python
#picking the best model
best_model_name = max(results, key=results.get)
print("Model accuracy comparison:")
for name, acc in results.items():
    print(f"  {name}: {acc:.4f}")

print(f"\nBEST MODEL: {best_model_name} (Accuracy = {results[best_model_name]:.4f})")
```
